```python
# Importaçãoes
import pandas as pd

# Carregar dados
url = "https://raw.githubusercontent.com/ect-info/ml/master/dados/preco_casas.csv"

df = pd.read_csv(url)

df.head()
```

```python
X = (df.iloc[:, 0].values).reshape(-1, 1) # Área -> Independente
y = df.iloc[:, 1].values # Preço -> Dependente

print(X[0:3])
print(y[0:9])
```

```python
# Importando o matplot
import matplotlib.pyplot as plt

# Plotando os dados
plt.plot(X, y, 'o')
plt.xlabel('Área')
plt.ylabel('Preço')
plt.show()
```

```python
# Importando o sklearn
from sklearn.linear_model import LinearRegression

# Criando o modelo
modelo = LinearRegression()
modelo.fit(X, y)
```

```python
# Plotando Resultados
y_predict = modelo.predict(X) # Gerando o valores de y por meio da regressão treinada para gerar a reta
plt.plot(X, y, 'o')
plt.plot(X, y_predict, color='red')
plt.xlabel('Área')
plt.ylabel('Preço')
plt.show()
```
