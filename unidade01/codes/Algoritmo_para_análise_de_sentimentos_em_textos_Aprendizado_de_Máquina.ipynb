# Importação dos dados de análise

```python
# Bibliotecas usadas
import pandas as pd

# Importando base de dados
url = 'https://raw.githubusercontent.com/ect-info/ml/master/dados/Restaurant_Reviews.tsv'

df = pd.read_csv(url, sep='\t', quoting = 3)
df.head()
```

# Importando e configurando algumas bibliotecas importantes
- Lib. Natural Language Toolkit (nltk) → Possuem modelo pré-treinados para análise de sentimentos

- Lib. googletrans-py → Para a tradução do português para o inglês

```python
# Instalando o googletrans-py
!pip install googletrans-py
```

```python
# Importando o NLTK e seu modelo pré-treinado
import nltk
nltk.download('vader_lexicon')
from nltk.sentiment.vader import SentimentIntensityAnalyzer

# Importando o Google Translator
from googletrans import Translator
translator = Translator()
```

## Fazendo testes com texto em Português

```python
# Textos
listTextoPTBR = [
    "Oi. Como você está?",
    "Bom dia.",
    "Bom dia!",
    "Quero que o mundo de exploda hoje. Estou sem paciência para tudo!",
    "Que RAIVA! Todo mundo é idiota! Não gosto de pessoas!"
]
```

```python
# Traduzindo texto para o inglês
listTextoEN = []
for texto in listTextoPTBR:
  listTextoEN.append(translator.translate(texto, dest='en').text)

# Mostrando texto traduzidos
for texto in listTextoEN:
  print(f"- {texto}")
#
```

```python
# Fazendo a análise de sentimento desses textos
sid = SentimentIntensityAnalyzer()
for texto in listTextoEN:
  s = sid.polarity_scores(texto)
  print(f' -> Frase: "{texto}":')
  print(f"    - Percentual negativo: {s['neg'] * 100:.2f}%")
  print(f"    - Percentual positivo: {s['pos'] * 100:.2f}%")
  print(f"    - Percentual neutro: {s['neu'] * 100:.2f}%")
  print(f"    - Score Composto: {s['compound']:.4f}")

  # Determinando o sentimento geral
  if s['compound'] >= 0.05:
    sentiment = 'Positivo'
  elif s['compound'] <= -0.05:
    sentiment = 'Negativo'
  else:
    sentiment = 'Neutro'
  print(f"    - Sentimento: {sentiment}\n")
```

## Analisando frases da base de dados

```python
# Analisando a base de dados
# Pegando as 5 primeiras frases da base de dados
frases_analise = df['Review'][0:5]

# Analisando o sentimento e mostrando
for frase in frases_analise:
  s = sid.polarity_scores(frase)
  print(f' -> Frase: "{frase}":')
  print(f"    - Percentual negativo: {s['neg'] * 100:.2f}%")
  print(f"    - Percentual positivo: {s['pos'] * 100:.2f}%")
  print(f"    - Percentual neutro: {s['neu'] * 100:.2f}%")
  print(f"    - Score Composto: {s['compound']:.4f}")

  # Determinando o sentimento geral
  if s['compound'] >= 0.05:
    sentiment = 'Positivo'
  elif s['compound'] <= -0.05:
    sentiment = 'Negativo'
  else:
    sentiment = 'Neutro'
  print(f"    - Sentimento: {sentiment}\n")
```

```python
# Classificando todos e gerando uma tabela com a classificação
for frase in df['Review']:
  s = sid.polarity_scores(frase)

  # Definindo o sentimento da frase no dataframe
  if s['compound'] >= 0.05:
    df.loc[df['Review'] == frase, 'sentiment'] = 'Positivo'
  elif s['compound'] <= -0.05:
    df.loc[df['Review'] == frase, 'sentiment'] = 'Negativo'
  else:
    df.loc[df['Review'] == frase, 'sentiment'] = 'Neutro'

# Mostrando DataFrame
df.head()
```
